# Phase 1: EDA Findings
Based on our exploratory data analysis, here are the key findings about noise and formatting patterns:
1. **Class Balance**: About 123k S1 entities are singletons (0 matches). Most have 2 to 5 matches.
2. **Missing Data**: S1 is complete, but S2 and S3 have ~3% missing addresses.
3. **Country Distribution**: France only appears in the test set.
4. **Name Variations**:
   - OCR/Typo errors: `5` instead of `S`, `0` instead of `o`.
   - Legal suffixes: `Corp` vs `Corporation`, `Pvt Ltd` vs `Private Limited`, `L.L.C.` vs `LLC`, `Inc` vs `Incorporated`.
   - Extra noise: `(Office)`, `[Consultancy]`, URLs (`cardiologymetrocare.com`).
   - Regional scripts: Hindi and Telugu text appearing in names/addresses (e.g., `??????? ???? ???????? ???????`).
5. **Address Variations**:
   - Reordering: `TX, 158 Simpson Lane, Somerset` vs `158 Simpson Lane, Somerset, Texas`.
   - Abbreviations: `Circle` -> `CIR`, `Drive` -> `DR`, `Avenue` -> `AVE`, `Road` -> `RD`, `Street` -> `ST`.
   - State abbreviations: `Maharashtra` -> `MH`, `Texas` -> `TX`, `Tennessee` -> `TN`, etc.
